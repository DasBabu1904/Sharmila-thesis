```python
import pandas as pd
import numpy as np
```

```python
# Load data
df = pd.read_excel('perfect_final_data_for_multilevel_analysis.xlsx')

# Display basic info
print(df.shape)
print(df.columns.tolist())
print(df.info())
print(df.head())
```

```python
df2 = df[df['V447'] == 0].copy()
df2.shape
```

```python
df = df.drop(columns=['V447'])
```

```python
feature_cols = [
    'V012', 'V024', 'V025', 'V106',
    'Drinking_water', 'Religion_status', 'Media_acccess', 'Wealth_Status',
    'Number_of_children', 'Living_children', 'Birth_Interval',
    'contraceptive_use', 'husband_education', 'profession_of_husband',
    'Age_of_husband', 'Respondent_Age_first_birth'
]

# Make sure all columns exist
existing_cols = [col for col in feature_cols if col in df.columns]
missing = set(feature_cols) - set(existing_cols)
if missing:
    print("Missing columns:", missing)
    # You'll need to create them or drop them
    # For now, we'll use only existing ones
    feature_cols = existing_cols

X = df[feature_cols].copy()
y = df['BMI'].copy()   # categorical target

# Verify target values
print(y.value_counts().sort_index())
```

```python
print(X.isnull().sum())
```

# encoding for nominal categories 

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder

# 1. Split first
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)


# 2. Identify nominal columns
nominal_cols = ['V024', 'Drinking_water', 'profession_of_husband']

# 3. Fit encoder on training only
encoder = OneHotEncoder(drop='first', handle_unknown='ignore', sparse_output=False)
encoder.fit(X_train[nominal_cols])

# 4. Transform both train and test
X_train_encoded = encoder.transform(X_train[nominal_cols])
X_test_encoded = encoder.transform(X_test[nominal_cols])

# 5. Drop original nominal columns and concatenate encoded ones
X_train = X_train.drop(columns=nominal_cols).reset_index(drop=True)
X_test = X_test.drop(columns=nominal_cols).reset_index(drop=True)

# Convert encoded arrays to DataFrames with proper column names
train_enc_df = pd.DataFrame(X_train_encoded, columns=encoder.get_feature_names_out(nominal_cols))
test_enc_df = pd.DataFrame(X_test_encoded, columns=encoder.get_feature_names_out(nominal_cols))

X_train = pd.concat([X_train, train_enc_df], axis=1)
X_test = pd.concat([X_test, test_enc_df], axis=1)
```
